```python
from pathlib import Path
import os

# Define the data root directory (repo-relative)
DATA_ROOT = Path("/home/prasndhdl/Downloads/avatar/Avatar-PrashantD_Forked/Data_clean")
# Label aliases to handle different naming conventions
LABEL_ALIASES = {
    "backward": {"backward"},
    "forward": {"forward"},
    "land": {"land", "landing"},
    "landing": {"land", "landing"},
    "left": {"left"},
    "right": {"right"},
    "takeoff": {"takeoff"},
}

def get_label_csv_paths(label):
    """
    Find all CSV files for a given movement label under Data_clean.
    
    Parameters:
        label (str): The movement label (e.g., "Forward", "Backward", "land")
    
    Returns:
        list: Sorted list of Path objects pointing to CSV files
    """
    # Get all matching aliases for the label
    aliases = LABEL_ALIASES.get(label.lower(), {label.lower()})
    
    # Find all directories that contain any of the aliases in their name
    directories = sorted(
        path
        for path in DATA_ROOT.rglob("*")
        if path.is_dir() and any(alias in path.name.lower() for alias in aliases)
    )
    
    # Collect all CSV files from these directories
    csv_files = []
    for directory in directories:
        csv_files.extend(sorted(directory.glob("*.csv")))
    
    return sorted(csv_files)
```

```python
import pandas as pd

def load_exg_16(file_path):
    # Try common formats (tab, comma, auto)
    for read_kwargs in [
        {"sep": "\t", "engine": "python"},
        {"sep": ",", "engine": "python"},
        {"sep": None, "engine": "python"},  # auto-detect
    ]:
        try:
            df = pd.read_csv(file_path, header=None, on_bad_lines="skip", **read_kwargs)
            if df.shape[1] >= 17:
                # many BrainFlow files: col0 is sample index, EXG in 1..16
                return df.iloc[:, 1:17].copy()
            if df.shape[1] >= 16:
                # fallback if file already has only 16 channels
                return df.iloc[:, :16].copy()
        except Exception:
            pass
    return None

csv_files = get_label_csv_paths("Land")
good_dfs = []
bad_files = []

for f in csv_files:
    exg = load_exg_16(f)
    if exg is None:
        bad_files.append(str(f))
    else:
        good_dfs.append(exg)

print(f"Landing files total: {len(csv_files)}")
print(f"Loaded successfully: {len(good_dfs)}")
print(f"Failed: {len(bad_files)}")

if bad_files:
    print("\nFirst 10 failed files:")
    for bf in bad_files[:10]:
        print(" ", bf)

concatenated_df_land = pd.concat(good_dfs, ignore_index=True)
concatenated_df_land.columns = [f"EXG Channel {i}" for i in range(16)]
print(concatenated_df_land.shape)
##(One file is partially malformed (header/noise lines mixed into raw data), but we can still load the valid EXG channels and ignore the bad lines.
```

```python
import numpy as np
import pandas as pd

desired_rows = 1250
total_files_processed = 0
skipped_files = 0
processed_dfs = []

for filepath in get_label_csv_paths("Land"):
    df = load_exg_16(filepath)  # uses your robust parser
    if df is None or df.empty:
        skipped_files += 1
        print(f"Skipped (unreadable): {filepath}")
        continue

    num_rows = len(df)

    if num_rows < desired_rows:
        padding_rows = desired_rows - num_rows
        padding_df = pd.DataFrame(np.zeros((padding_rows, 16)))
        df = pd.concat([df, padding_df], ignore_index=True)
        print(f"{filepath.name}: Padded {padding_rows} rows (before: {num_rows})")
    elif num_rows > desired_rows:
        truncated_rows = num_rows - desired_rows
        df = df.iloc[:desired_rows].copy()
        print(f"{filepath.name}: Truncated {truncated_rows} rows (before: {num_rows})")
    else:
        print(f"{filepath.name}: Already {desired_rows} rows")

    processed_dfs.append(df)
    total_files_processed += 1

concatenated_df_Land = pd.concat(processed_dfs, ignore_index=True)
concatenated_df_Land.columns = [f"EXG Channel {i}" for i in range(16)]

print(f"\nProcessed: {total_files_processed}, Skipped: {skipped_files}")
print(f"Final shape: {concatenated_df_Land.shape}")
```
